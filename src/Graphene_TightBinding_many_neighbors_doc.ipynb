```python
import sys
from scipy.constants import eV

sys.path.append("/home/lplaja/crystal_project/src")

import numpy as np
import crystal as cr
import grid as grid
from Graphene_TightBinding_many_neighbors import TB_graphene_CMCP, qe, epsk, hbar
from Field import polarizedElectricField, lambda2T, env_sin2

# Create an hexagonal grid
filter=grid.polygonfilter
filter_args={'nsides':6, 'radius': 2/3}
ggr=grid.UniformCartesianGrid(2,[[-0.75,0.75],[-0.75,0.75]],[500,500], origin=(0.0,0), filter=filter, filter_args= filter_args)

# Create the graphene crystal
gcr=cr.crystal(cr.species['graphene'], grid=ggr, space='reciprocal')

# Create the Field
lambda0=3000e-9
T0=lambda2T(lambda0)
#npt=1024*32
npt=1024*16
tt=grid.UniformCartesianGrid(1,limits=[0,8*T0], nptx=npt)

Efield=polarizedElectricField(tt,I_W__cm2=5e10,lambda0_nm=lambda0*1e9, env=env_sin2, phi_rad=-np.pi/2, chi_rad=np.pi/2,ellip=0)

# Create the TightBinding object
gtb=TB_graphene_CMCP(gcr, Efield)
print(gtb.orbital_names)

print('neighbor_hoppings/eV')
print(f'{gtb.neighbor_hoppings.shape=}')

print(f'From orbital {gtb.orbital_names[0]}')
for m, ineighbor_hoppings in enumerate(gtb.neighbor_hoppings[0]):
    print(f'\t to orbital {gtb.orbital_names[m]}')
    for j, jhopping in enumerate(ineighbor_hoppings):
         print(f'\t \t at {gtb.neighbor_positions[0,m,j,:]}    hopping: {jhopping/eV}')

print()
print(f'From orbital {gtb.orbital_names[1]}')
for m, ineighbor_hoppings in enumerate(gtb.neighbor_hoppings[1]):
    print(f'\t to orbital {gtb.orbital_names[m]}')
    for j, jhopping in enumerate(ineighbor_hoppings):
         print(f'\t \t at {gtb.neighbor_positions[1,m,j,:]}    hopping: {jhopping/eV}')

print('-------------------------------------')
print()
print(gtb.tnm(0.3,0.5)/eV)

```
